## The problem.
A drone company (RedKite Aerospace) needs its drones to be able to read hand signals from the person receiving the drone delivery. Different gestures need to map to different commands and the drone needs to be able to differentiate
between them. A segmentor that already exists on the aircraft will cut the relevant parts out (the gestures) and pass each one to the model as 5 frames.
<br>
The task at hand is a **13-class classification problem** over these pre-segmented frames: 5 frames go into the model, and it outputs one label for the given gesture.
<br>
The 13 classes (gestures) that exist in the dataset are:
- `AllClear`
- `HaveCommand`
- `Hover`
- `Land`
- `LandingDirection`
- `MoveAhead`
- `MoveDownward`
- `MoveToLeft`
- `MoveToRight`
- `MoveUpward`
- `NotClear`
- `SlowDown`
- `WaveOff`
<br>
I will be building a classifier model. There are some restrictions: the computer is already overloaded as it is running navigation, mapping, comms etc., so there is very little space and power to run a complicated classifier model. Due to that, we are restricted to only use a 2-D convolutional backbone. The model also has to run inference on one instance at a time on an embedded board, which essentially means that it needs to stay small enough for real-time operation to be possible.
<br>
A 2-D CNN by itself would not be able to achieve satisfactory accuracy results on such a small dataset. Using an existing ImageNet-pretrained backbone could be a solution - that way the model doesn't have to learn all the visual features from scratch. We can also apply data augmentation, which would help us get more use out of the small dataset that we have.
<br>
Another potential architectural issue is that a 2-D CNN is only able to take 1 image as input to produce 1 output label, so normally it doesn't have a way to use the sequence of 5 frames that we have.
Getting the 5 frames into a 2-D CNN is an issue that will need to be addressed by attempting various appraoches to this problem.

1. **Singular frame** - classify simple based on one frame out of 5, which is the simplest possible approach. This will serve as the *baseline*.
2. **Early frame fusion** - we will stack the 5 frames into one as channels - this is a workaround for the network to get all 5 images in one input.
3. **Late frame fusion** - we will run the same backbone on each of the five frames in the sequence and then we will combine the resulting features in the head.
<br>
The given dataset itself could prove to be a problem - it was recorded with RedKite's own staff. On the actual day of the demo, a member of the client's staff will actually be doing the gestures for testing, which will be a person that the model has never seen before in an unseen location.

I will be building a classifier model. There are some restrictions to work around: the onboard computer is already overloaded running navigation, mapping and comms, so there isn't much space or power left for the classifier. The spec restricts this to 2-D convolutional backbones only, so no RNN/LSTM/GRU, no 3-D convolutions and no temporal-attention or video-transformer architectures. The model also has to run inference on one instance at a time on an embedded board, so it needs to stay small enough that real-time operation is actually plausible, and I have to argue that rather than just assume it. <br>
ImageNet-pretrained backbones and standard augmentation are allowed, but pretraining on any external gesture or action-recognition dataset is not.

A 2-D CNN on its own might not get far on a dataset this small, but using an ImageNet-pretrained backbone means the model doesn't have to learn general visual features from scratch, and augmentation should squeeze a bit more out of the limited data we have. Picking a small backbone like ResNet-18 also keeps the parameter count and the per-instance inference cost low enough that real-time operation on the board stays plausible. <br>
The other problem is that a 2-D CNN takes one image and gives one output, so it has no way of using the fact that an instance is five frames in a sequence. That information has to come in through how I feed the frames to the network or through a head that I put on top of the backbone, since the backbone itself can't do it.

So the main design question this notebook is built around is how to get all five frames into a 2-D CNN. A few approaches will be developed and compared:
1) **Single-frame** - classify one frame and ignore the other four. This is a baseline to compare against rather than a real candidate, since it throws away most of the instance.
2) **Early fusion** - stack the five frames together as channels so the network gets all of them in one input.
3) **Late fusion** - run the same backbone over each of the five frames and combine the resulting features with a head that I design and train myself.

The dataset is another thing worth flagging early. It was recorded in a single afternoon with the company's own staff standing in front of a camera, and none of them are trained signallers. On the demo day it will be a member of the client's staff doing the gestures, so it will be a person the model has never seen, most likely in a different spot with different lighting. That gap between who is in the training data and who the model actually has to work for is the thing that shapes the evaluation framework in Step 3, and it comes up again at the end.
